# Derivatives

Derivatives are fundamental concepts in calculus that measure how a function changes as its input changes. Derivatives are used in areas of machine learning like optimization algorithms, training models and improving the performance of various machine learning techniques.

Derivatives are used to understand how changes in model parameters affect the model's performance, typically measured by a **loss function**.

#### **Applications of derivatives in machine learning:**

1. **Gradient descent optimization:** is an optimization algorithm used to minimize the loss function of a machine learning model. The loss function quantifies the difference between the predicted and actual values. Derivatives, specifically gradients, indicate the direction and rate of change of the loss function with respect to the model parameters $(w, b, etc..)$
    - Gradient calculation: The gradient of the loss function is a vector of partial derivatives. It points in the direction of the steepest increase of the function. By moving in the opposite direction of the gradient, the algorithm iteratively reduces the loss, increasing the model's performance.
    - Learning rate: The learning rate determines the step size during each iteration. A small learning rate results in a slow convergence, while a large learning rate can cause overshooting the **Global or Local Minimum**.
    - Iterate: Repeat the process until convergence to the **Global or Local Minimum**.
  
2. **Backpropagation in Neural Networks:** Backpropagation is an algorithm that uses derivatives to update the network's parameters and minimize the loss. It works by propagating the error signal backward through the network, from the output layer back to the input layer.
   - Forward propagation: Input data flows through the network, each neuron calculates its output based on weights, biases, and activation functions.
   - Loss Calculation: The network's final output is compared to the ground truth, resulting in a loss value.
   - Backward propagation: Then we go backward through the network from the last layer to the input layer. Using the [chain rule](./chain_rule.ipynb) of calculus, it calculates the derivative (gradient) of the loss with respect to each weight and bias in the network. The chain rule allows the algorithm to determine how much each parameter $(w, b)$ in every layer contributed to the overall error.
   - Gradient descent: The weights and biases are updated in the opposite direction of their respective gradients.

3. **Regularization Techniques:** are used to prevent overfitting by adding a penalty term of the loss function. Derivatives are used to compute the gradients of these regularized loss functions, ensuring that the penalty terms are incorporated into the optimization process.
   - L2 Regularization: Reduces model complexity and prevents overfitting by penalizing large weights.
   - L1 Regularization: Adds a penalty proportional to the absolute value of the weights. This can shrink some weights to exactly zero, effectively performing feature selection by removing unimportant features from the model.

4. **Other applications include Hyperparameter Optimization, designing activation functions, and more..**

#### **Slope On A Curve:**

The slope is constantly changing, unlike the linear line.

<p align="center">
    <img src="./showcase_images/derivative_2.png" alt="derivative graph" width="400">
</p>

$$\text{Average Rate of Change} = \frac{f(x + h) - f(x)}{\cancel{x} + h - \cancel{x}} = \frac{f(x + h) - f(x)}{h}$$
- The denominator $x$'s cancel out.
- **Secant Line:** A line that intersects two or more points on a curve.

**Now let's make $h$ as close to zero as possible, i.e., moving the $x+h$ point closer to the $x$ point.**
- Note: This concept of moving two points infinitely close together to find the slope at a single point is the core idea that powers gradient-based optimization in ML.

- **Note:** While it looks like points $\color{blue} x$ and $\color{blue} x+h$ are at the same coordinates, they are not, $\color{blue} x+h$ is just a little bit more along the curve, but still very close to zero.
  - Examples: 
    - $x = 2$, $h=0.001$, $x + h = 2.001$
    - $x = 2$, $h=0.00001$, $x + h = 2.00001$, $h$ is closer to zero here than the above example.
  
- Now we call the secant line, the **Tangent Line** (point of tangency), which is a line that just touches the curve at a point, matching the curve's slope at that point.

  - As $h$ approaches zero, the secant line pivots to become the Tangent line.
  - A tangent line is the limit ($\lim_{h \to 0}$) of a secant line as the two points $x$ and $x+h$ on the curve get infinitely close to each other, however, $h$ won't ever equal $0$ (which would cause a division by zero).

  - The tangent line is the best straight-line approximation of the curve at that specific point. It captures the **instantaneous rate of change** or **Derivative**, which is the slope of the curve at that point.
    - The **Derivative** tells us the slope at a specific point.

--- 

$$\text{Derivative} = \frac{dy}{dx} = f'(x) = y' = \dot{y} = \lim_{h \to 0} \frac{f(x+h) - f(x)}{(\cancel{x}+h) - \cancel{x}}$$


$$= \lim_{h \to 0} \frac{f(x+h) - f(x)}{h}$$

- Where:
  - $y'$ means $y$ prime
    - $\text{Derivative} = \frac{dy}{dx} = f'(x) = y' = \dot{y}$ are different ways this function might be noted.

  - $f(x)$ is the value of the $y$-coordinate for a given $x$-coordinate. For a point like $(1,10)$, this means $x = 1$ and $f(1) = 10$

  - When dealing with tangent lines we use $dy$ and $dx$ instead of $\Delta y$ & $\Delta x$.
    - $\Delta$ is called delta.
    - $\Delta y$ & $\Delta x$ represents the slope of the secant line (an average change).
    - $dy$ & $dx$ represents the slope of the tangent line (the instantaneous change).
      - They both mean, the rate of change of a vertical variable $y$, with respect to a horizontal variable $x$.
      - Both describe the rate of change of $y$-coordinate with respect to $x$-coordinate, but $\Delta$ refers to the average change over an interval, while $d$ refers to the instantaneous change at a single point.

  - The denominator $(x+h) - x$, the $x$'s cancel out, $h$

  - $\lim_{h \to 0}$ means as the point $h$ approaches $0$ (i.e., the point $x+h$ gets closer to the point $x$, on the curve), we want to make the $h$ as close as zero as possible.

#### Derivative Example:

**What is the slope of the tangent line to $f(x) = x^2$ at point $(1, 1)$?**

<p align="center">
    <img src="./showcase_images/derivative_6.png" alt="derivative graph" width="200">
</p>

$$= \lim_{h \to 0} \frac{f(x+h) - f(x)}{h}$$

Add the $x^2$

$$= \lim_{h \to 0} \frac{(x+h)^2 - (x)^2}{h}$$

$$= \lim_{h \to 0} \frac{(x+h) (x+h) - (x)^2}{h}$$


**Simplify** $(x+h)(x+h)$ use the below square:
<p align="center">
    <img src="./showcase_images/derivative_7.png" alt="derivative graph" width="100">
</p>

$$\text{Result} = x^2 + hx + hx + h^2$$

$$ = x^2 + 2hx + h^2$$

Plug back in to the formula:

$$= \lim_{h \to 0} \frac{\cancel{x^2} + 2hx + h^2 \cancel{- (x)^2}}{h}$$


$$= \lim_{h \to 0} \frac{ 2hx + h^2}{h}$$

$$= \lim_{h \to 0} \frac{ \cancel{h} (2x + h)}{\cancel{h}}$$

$$= \lim_{h \to 0} 2x + h$$

Plug in zero for the h:

$$= 2x + 0$$

$$= 2x $$

$$f'(x) = 2x $$

The derivative is equal to $2x$ or in other words the slope of the tangent line is $2x$.

Now we can plug in the $1$ from the **x-coordinate** of the point $(1, 1)$, i.e., $(x,y)$.


$$f'(1) = 2(1) = 2$$

**So the slope of the tangent line at the point $(1, 1)$ is $2$.**
- This means that the tangent line at point $(1, 1)$ on the parabola $f(x)=x^2$ has a 'rise over run' of 2. For every 1 unit you move right on the tangent's $x$-axis, the line rises by 2 units on the $y$-axis.

## **Overall Big Picture**

How do we use derivatives in machine learning?


We use derivatives to find the **Local** or **Global minimum**.
- Local minimum is the lowest point of a function within a specific, small neighborhood.
- Global minimum also known as an absolute minimum, is the lowest overall value of a function across its entire domain.

- Example: The local and global minimums on our 2D derivative example above will be located at the lowest point of the parabola.

**Difference between 2D vs 3D+**
- In 2D, the derivative told you the slope at a single point. But on a 3D surface, the "slope" depends on which direction you are facing. If you're on the side of a hill, the slope is steep if you face uphill, but zero if you walk horizontally along the hill's side.

- So, we need a concept to tell us not just how steep the derivative is, but also in which direction it's steepest. This is the **Gradient**
  - The **gradient** is a derivative of a function that has more than one input variable.
    - "The gradient is the generalization of the derivative to multivariate functions. It captures the local slope of the function, allowing us to predict the effect of taking a small step from a point in any direction."
      - The small step is us tweaking the weights and biases so that the loss function is lower, giving us a better model, and reaching the local or global minimums.

    - The gradient is a vector (think of it as an arrow) that points in the direction of the steepest **ascent** at any given point on the surface. It's calculated using **partial derivative**. You essentially take the derivative with respect to each parameter (weights $w$, bias $b$, etc..) individually, and bundle them together.
      - The derivative in 2D space is a single number telling us the slope at a certain point on a curved or linear line.
      - The derivative in 3D space is a **vector** that points in the direction of the steepest ascent.

**Key Takeaways:**
- Derivatives measure instantaneous change(slope).
- Gradients extend derivatives to multi-parameter (weights and biases, etc..) functions.
