# Batch Gradient Descent (BGD)


Gradient descent is the backbone of how models learn. It attempts to minimize the cost function of a model by iteratively adjusting the models learned parameters weights $(w)$ and biases $(b)$, to reduce the difference between predicted and actual values, to improve the model's performance.

Steps:
- **Backpropagation** uses the [chain rule](../../Resouces/chain_rule.ipynb) to computes the **gradients** of the loss function with respect to each parameter $(w,b)$, across all layers in the network.
- Gradients are then used by Batch Gradient Descent to update the parameters layer-by-layer, moving toward minimizing the loss function.
