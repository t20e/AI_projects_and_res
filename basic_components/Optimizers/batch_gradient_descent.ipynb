# Batch Gradient Descent (BGD)


Batch Gradient Descent also called **Gradient Descent** attempts to minimize the **cost** function of a model by iteratively adjusting the models learned parameters weights $(w)$ and biases $(b)$, to reduce the difference between predicted and actual values, to improve the model's performance.

The "Batch" in BGD means that the gradients are calculated by summing or averaging the loss error across the entire training dataset before making a single update to the learned parameters.

Steps:
- **Backpropagation** uses the [chain rule](../../Resouces/chain_rule.ipynb) to compute the **gradients** of the cost function with respect to each parameter $(w,b)$, across all layers in the network.
- Gradients are then used by Batch Gradient Descent to update the parameters layer-by-layer, moving toward **minimizing** the **cost** function.

**General parameter update formulas:**

The formulas below must be applied to each weight and bias in a network. 

$$ w = w - \alpha * \frac{\partial C(w)}{\partial w} $$

- $\frac{\partial C(w)}{\partial w}$ Is the derivative (the gradient) of $C$ (cost function) with respect to $w$, i.e., 'how much does $C$ change when $w$ changes'.

- $\alpha$ symbol is called 'alpha', which is the learning rate, i.e., how much to change the parameter $w$ by per a single training iteration. This is a small arbitrary float (e.g., 0.1, 0.01, etc...)

$$ b = b - \alpha * \frac{\partial C(b)}{\partial b} $$

- $\frac{\partial C(b)}{\partial b}$ is the derivative (the gradient) of $C$ with respect to $b$.



Backpropagation computes the gradient of the Cost function with respect to each weight and bias in the network using the chain rule.

```python
import numpy as np

def update_parameters(w, b, dw, db, learning_rate=0.1):
    """
    Performs a single batch gradient descent update step. This is done for all weights and bias in the network.

    Args:
        w (np.array): Current weights
        b (float): Current bias
        dw (np.array): Gradients of the cost with respect to w
        db (float): Gradients of the cost with respect to b
        learning_rate (float): The learning rate (alpha)
    Returns:
        tuple: (w_new, b_new) containing the updated parameters
    """
    w = w - learning_rate * dw
    b = b - learning_rate * db
    return w, b
```
