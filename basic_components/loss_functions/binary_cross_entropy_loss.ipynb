# Binary Cross Entropy Loss

- For binary classification problems, where a model predicts a $0$ or  $1$, true or false, cat or no-cat, we use a binary cross entropy. It quantifies the difference between the actual class labels (0 or 1) and the predicted probabilities output by the model. The lower the binary cross-entropy value, the better the model’s predictions align with the true labels.

$$BCE = -\frac{1}{N} \sum^N_{i=1} [y_i log(p_i) + (1 - y_i) log(1-p_i)]$$

Where:
* $N$ is the number of samples.
* $y_i$ is the target binary label ($0$ or $1$) of the $i^{th}$ sample.
* $p_i$ is the predicted probability of the $i^{th}$ sample.

```python
import numpy as np
def BCE(y_true, y_pred):
    """
    Computes Binary Cross Entropy

    Args:
        y_true: The target binary label vector (0 or 1).
        y_pred: The model's predicted probability vector.
    
    """
    epsilon = 1e-15 # Small number to prevent division by zero.
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon) # Clip the predictions so that the if the prediction is exactly 0 or 1 it won't break the bce function.

    bce = -np.mean(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))
    print("Loss score:" , bce)
```

**Show a poor score, where the models predictions weren't good**

In the example below:
* `y_true = [0, 1, 1, 0, 1]` represents five true labels samples. For example, if $0$ means "cat" and $1$ means "dog," this vector represents `["cat", "dog", "dog", "cat", "dog"]`.

* `y_pred = [0.9, 0.2, 0.3, 0.2, 0.7]` represents the model's predicted probabilities. For instance, the model predicts the first sample is a dog with a probability of $0.9$. This prediction is poor because the true label is a cat ($0$).

```python
# Example true labels and predicted probabilities
y_true = np.array([0, 1, 1, 0, 1])
y_pred = np.array([0.9, 0.2, 0.3, 0.2, 0.7])
y_true, y_pred
```

```python
BCE(y_true, y_pred)
```

**Show a better score**
- The closer the predictions are to the target the **lower** the loss score.

```python
# Example true labels and predicted probabilities
y_true = np.array([0, 1, 1, 0, 1])
y_pred = np.array([0.1, 0.9, 0.8, 0.2, 0.7])
y_true, y_pred
```

```python
BCE(y_true, y_pred)
```
