# Categorical Cross-Entropy

Categorical Cross-Entropy is a loss function used for** multi-class classification** problems, where a model outputs a prediction for every class in the dataset.

Use-Case-Example:
- We have a dataset of images, and each image can have one of these animals -> ["dog", "cat", "bird", "snake"]. So we have 4 classes for this dataset.
- A single sample would be one image, and its corresponding label.
  - Example: one dog image, and its one-hot encoded label -> [1,0,0,0]
- The output layer of the network typically has **multiple neurons**, one for each class, and uses a **softmax** activation function to output a probability distribution over the classes.

$$L(y, \hat{y}) = - \sum^C_{i=1} y_i log(\hat{y_i})$$

Where:
* $y$ is a binary indicator (0 or 1) representing the true class label from a from a one-hot encoded vector.
* $\hat{y}$ is the predicted probability for class $i$.
* $C$ is the number of classes.

```python
import numpy as np

def CCE(y_true, y_pred):
    """
    Computes the Categorical Cross-Entropy loss for a single sample.

    Args:
        y_true (np.array): A one-hot encoded vector of the true label.
        y_pred (np.array): The model's predicted probabilities for each class.
    """
    # Add a small epsilon for numerical stability
    epsilon = 1e-15
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)

    # The formula is the negative sum of the element-wise multiplication of y_true and the log of y_pred.
    cce_loss = -np.sum(y_true * np.log(y_pred))
    print("Loss score for a single sample:", cce_loss)
```

- Example bad score: The model is predicting that theres a cat in the image but its actually a bird.

```python
#                 ["dog", "cat", "bird", "snake"]
y_true = np.array([0, 0, 1, 0])
y_pred = np.array([0.02, 0.95, 0.01, 0.02])
CCE(y_true=y_true, y_pred=y_pred)
```

- Example a good score: The model correctly predicts that theres a snake in the image.
  - **The lower the score the better the result.**

```python
#                 ["dog", "cat", "bird", "snake"]
y_true = np.array([0, 0, 0, 1])
y_pred = np.array([0.02, 0.02, 0.01, 0.9])
CCE(y_true=y_true, y_pred=y_pred)
```
